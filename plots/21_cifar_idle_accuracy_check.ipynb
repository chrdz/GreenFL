```python
import numpy as np
import pandas as pd

# ============================================================
# 1) CIFAR reference configuration (from your LaTeX reference)
# ============================================================

# End time for fine-tuning (s) columns for CIFAR
S_LIST = [20, 40, 60, 80, 100]

# Baseline + budget strings must match the reference table exactly.
# Keyed by training time T.
CIFAR_BASELINE_BY_T = {
    90: {"budget_str": r"20.78 / 86.04\%", "baseline_mean": 74.73, "baseline_std": 0.21},
    75: {"budget_str": r"15.54 / 64.34\%", "baseline_mean": 74.19, "baseline_std": 0.17},
    60: {"budget_str": r"11.73 / 48.58\%", "baseline_mean": 73.30, "baseline_std": 0.18},
    45: {"budget_str": r"8.84 / 36.6\%",  "baseline_mean": 72.09, "baseline_std": 0.11},
    30: {"budget_str": r"6.2 / 25.65\%",  "baseline_mean": 68.69, "baseline_std": 0.51},
    15: {"budget_str": r"3.03 / 12.56\%", "baseline_mean": 62.13, "baseline_std": 0.40},
     5: {"budget_str": r"0.94 / 3.9\%",   "baseline_mean": 49.89, "baseline_std": 0.52},
     3: {"budget_str": r"0.56 / 2.31\%",  "baseline_mean": 44.05, "baseline_std": 0.30},  # only appears in ft=1
}

# Row ordering to match your reference table
ROW_ORDER_FT1 = [90, 75, 60, 45, 30, 15, 5, 3]
ROW_ORDER_FT3 = [90, 75, 60, 45, 30, 15, 5]


# ============================================================
# 2) Helpers: parsing + formatting (variance -> std)
# ============================================================

def _to_list(x):
    """Convert scalar or list-like cell to a flat list of floats."""
    if x is None:
        return []
    if isinstance(x, float) and np.isnan(x):
        return []
    if isinstance(x, (list, tuple, np.ndarray, pd.Series)):
        out = []
        for v in x:
            if isinstance(v, (list, tuple, np.ndarray, pd.Series)):
                out.extend(list(v))
            else:
                out.append(v)
        return [float(v) for v in out if v is not None and not (isinstance(v, float) and np.isnan(v))]
    return [float(x)]

def _fmt_num(x, decimals=2):
    """Format number like LaTeX table: 2 decimals, strip trailing zeros/dot."""
    if x is None:
        return ""
    if isinstance(x, float) and np.isnan(x):
        return ""
    s = f"{float(x):.{decimals}f}"
    return s.rstrip("0").rstrip(".")

def _tightcell(mean, std, bold=False):
    """Return LaTeX \\tightcell{mean}{std}, optionally bolding mean."""
    m = _fmt_num(mean, 2)
    s = _fmt_num(std, 2)
    if m == "":
        return "---"
    if bold:
        m = f"\\textbf{{{m}}}"
    if s == "":
        s = "0"
    return f"\\tightcell{{{m}}}{{{s}}}"

def _budget_cell(T):
    """Return CIFAR reference budget string, e.g. '20.78 / 86.04\\%'."""
    return CIFAR_BASELINE_BY_T.get(int(T), {}).get("budget_str", "")

def _baseline_cell(T):
    """Return CIFAR reference baseline \\tightcell{mean}{std}."""
    b = CIFAR_BASELINE_BY_T.get(int(T))
    if not b:
        return "---"
    return _tightcell(b["baseline_mean"], b["baseline_std"], bold=False)


# ============================================================
# 3) Aggregate your DF: (T,s) -> (mean_acc, std_acc)
#     - mean_acc from max_test_acc (mean over list)
#     - std_acc from sqrt(mean(var_test_acc))  (variance -> std)
#     - optional auto scaling if accuracies are in [0,1]
# ============================================================

def aggregate_results(df, auto_scale_if_fraction=True):
    required = {"max_test_acc", "var_test_acc", "s", "T"}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"DataFrame is missing required columns: {missing}")

    # Detect if accuracies look like fractions in [0,1]
    scale = 1.0
    if auto_scale_if_fraction:
        sample_vals = []
        for v in df["max_test_acc"].head(50).tolist():
            sample_vals.extend(_to_list(v))
        if len(sample_vals) and np.nanmax(sample_vals) <= 1.0:
            scale = 100.0

    agg = {}
    for (T, s), g in df.groupby(["T", "s"], dropna=False):

        # Mean accuracy: mean over all values (supports lists)
        acc_vals = []
        for v in g["max_test_acc"].tolist():
            acc_vals.extend(_to_list(v))
        mean_acc = float(np.mean(acc_vals)) * scale if len(acc_vals) else np.nan

        # Variance -> std: average variances then sqrt
        var_vals = []
        for v in g["var_test_acc"].tolist():
            var_vals.extend(_to_list(v))
        std_acc = float(np.sqrt(np.mean(var_vals))) * scale if len(var_vals) else np.nan

        agg.setdefault(int(T), {})[int(s)] = (mean_acc, std_acc)

    return agg


# ============================================================
# 4) Generate one CIFAR subtable (blue style like reference)
# ============================================================

def generate_cifar_subtable_latex(df, row_order, caption, label=None, auto_scale_if_fraction=True):
    agg = aggregate_results(df, auto_scale_if_fraction=auto_scale_if_fraction)

    # Include any extra T values not in the reference order (append at end)
    extra_T = sorted(set(map(int, df["T"].unique())) - set(row_order))
    full_row_order = list(row_order) + extra_T

    body_lines = []
    for T in full_row_order:
        budget = _budget_cell(T)
        baseline = _baseline_cell(T)

        row_dict = agg.get(int(T), {})

        # Best among s columns for this row
        means = [row_dict[s][0] for s in S_LIST if s in row_dict]
        best = np.nanmax(means) if len(means) else np.nan

        row_cells = [budget, str(int(T)), baseline]

        for s in S_LIST:
            if s in row_dict:
                mean_acc, std_acc = row_dict[s]
                bold = (not np.isnan(best)) and (not np.isnan(mean_acc)) and (abs(mean_acc - best) < 1e-12)
                row_cells.append(_tightcell(mean_acc, std_acc, bold=bold))
            else:
                row_cells.append("---")

        body_lines.append(" & ".join(row_cells) + r" \\")

    s_header = " & ".join([rf"\textbf{{{s}}}" for s in S_LIST])
    label_line = rf"\label{{{label}}}" if label else ""

    # Build the subtable exactly like your reference (blue caption + blue rules)
    latex = rf"""
\begin{{subtable}}{{\linewidth}}
    \centering
    {{
    \color{{blue}}%
    \arrayrulecolor{{blue}}%
    {{\captionsetup{{font={{footnotesize,color=blue}}}}%
    \caption{{{caption}}}
    }}
    {label_line}
    \renewcommand{{\arraystretch}}{{1.2}}
    \setlength{{\tabcolsep}}{{2pt}}
    \begin{{tabular}}{{|c|c|c|{'c'*len(S_LIST)}|}}
    \hline
    \multirow{{2}}{{*}}{{\shortstack{{\textbf{{Carbon}} \\ \textbf{{Budget ($k$)}}}}}} &
    \multirow{{2}}{{*}}{{\shortstack{{\textbf{{Training}} \\ \textbf{{Time ($T$)}}}}}} &
    \multirow{{2}}{{*}}{{\shortstack{{\textbf{{No-Slack}} \\ \textbf{{Baseline}}}}}} &
    \multicolumn{{{len(S_LIST)}}}{{c|}}{{\textbf{{End time for fine-tuning ($s$)}}}} \\
    \cline{{4-{3+len(S_LIST)}}}
     &  &  & {s_header} \\
    \hline
{chr(10).join(["    " + line for line in body_lines])}
    \hline
    \end{{tabular}}
    \arrayrulecolor{{black}}
    }}%
\end{{subtable}}
""".strip()

    return latex


# ============================================================
# 5) Generate full CIFAR table* (two blue subtables, blue caption)
# ============================================================

def generate_full_cifar_table_latex(tmp_1_cifar, tmp_3_cifar,
                                   label="tab:combined_table_cifar10",
                                   auto_scale_if_fraction=True):
    main_caption = (
        r"Final test accuracy on the CIFAR-10 dataset for different carbon budget constraints ($k$) and "
        r"fine-tuning end times ($s$). The first column reports the absolute available budget and the "
        r"percentage relative to the full-budget reference. The second and third columns show the training "
        r"time ($T$) and the accuracy of the slack-agnostic baseline, respectively. The remaining columns "
        r"correspond to varying fine-tuning end times~$s$, evaluated for two fine-tuning durations "
        r"($t_{\mathrm{ft}} \!=\! 1$ and $t_{\mathrm{ft}} \!=\! 3$). Numbers in parentheses denote the standard "
        r"deviation across three seeds. Bold values indicate the best accuracy in each row."
    )

    sub1 = generate_cifar_subtable_latex(
        tmp_1_cifar,
        row_order=ROW_ORDER_FT1,
        caption=r"Test accuracy with fine-tuning duration $t_{\mathrm{ft}}=1$.",
        label=None,
        auto_scale_if_fraction=auto_scale_if_fraction
    )

    sub3 = generate_cifar_subtable_latex(
        tmp_3_cifar,
        row_order=ROW_ORDER_FT3,
        caption=r"Test accuracy with fine-tuning duration $t_{\mathrm{ft}}=3$.",
        label="tab:cifar10-carbon_budget-3ft",
        auto_scale_if_fraction=auto_scale_if_fraction
    )

    full = rf"""
\begin{{table*}}[h!]
    
    \centering
    {{\captionsetup{{font={{footnotesize,color=blue}}}}%
    \caption{{{main_caption}}}
    \label{{{label}}}
    }}
    \scriptsize
    \renewcommand{{\arraystretch}}{{1.05}}
    \setlength{{\tabcolsep}}{{1pt}}

{sub1}

    \vspace{{3em}}

{sub3}

\end{{table*}}
""".strip()

    return full


# ============================================================
# 6) Usage (in your notebook)
# ============================================================

# latex_cifar = generate_full_cifar_table_latex(tmp_1_cifar, tmp_3_cifar)
# print(latex_cifar)
#
# with open("combined_table_cifar10_generated.tex", "w") as f:
#     f.write(latex_cifar)
```

```python
results_df = results_df[['availability','final_test_accuracy', 'seed','lr']]
tmp = results_df.groupby(['availability']).final_test_accuracy.apply(np.vstack).to_frame().reset_index()
tmp['var_test_acc'] = tmp['final_test_accuracy'].apply(lambda x : x.var(axis=0))
tmp['max_test_acc'] = tmp['final_test_accuracy'].apply(lambda x: x.max(axis=0))
tmp = tmp[['availability', 'var_test_acc', 'max_test_acc']]
tmp["cb"] = tmp["availability"].str.extract(r"(\d+)cb").astype(int)
tmp["ft"] = tmp["availability"].str.extract(r"(\d+)ft").astype(int)
tmp["s"] = tmp["availability"].str.extract(r"(\d+)sl").astype(int)
```
